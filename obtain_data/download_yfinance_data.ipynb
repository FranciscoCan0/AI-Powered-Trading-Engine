```python
import yfinance as yf
import pandas as pd
import numpy as np
from IPython.display import display
from pathlib import Path
```

```python
historical_data_intervals = ["1m", "2m", "1h"]
```

```python
#Import Tickers of Interest
stock_tickers = ["AAPL", "NVDA", "AMZN", #TECH
               "AXP", "BRK-B", "JPM", #BANKING
               "JNJ", "PFE", "UNH", #HEALTHCARE
               "PG", "WMT", "KO", #RETAIL
               "GE", "CAT", "LMT", #INDUSTRIAL/AEROSPACE
               "XOM", "CVX", "SHEL", # ENERGY
               "SONY", "BABA", "NVO", # INTERNATIONAL
               "SPY", "VOO", "IVV", "QQQ"] #ETFs

ticker_background_data = {}

#Create dataframe of info on each stock ticker
for ticker in stock_tickers:
    ticker_object = yf.Ticker(ticker)

    #Obtain Company Data
    temp = pd.DataFrame.from_dict(ticker_object.info, orient="index")
    temp.reset_index(inplace=True)
    temp.columns = ["Attribute", "Recent"]
    ticker_background_data[ticker] = temp



#Clean backgound df information
ticker_background_data_df = pd.concat(ticker_background_data).reset_index()
ticker_background_data_df = ticker_background_data_df.drop(columns = ['level_1'])
ticker_background_data_df.columns = ["Ticker", "Attribute", "Value"]

pd.set_option("display.max_columns", None)

ticker_background_data_df
```

```python
trading_data_by_interval = {}


for interval in historical_data_intervals:

    ticker_trading_data = []

    for ticker in stock_tickers:

        ticker_object = yf.Ticker(ticker)

        # Obtain different granularities of trading data for different training stages
        temp = ticker_object.history(period="max", interval=interval)
        temp["Ticker"] = ticker

        if not temp.empty:
            ticker_trading_data.append(temp)

    print(f"Interval {interval} download complete.")


    trading_data_by_interval[interval] = pd.concat(ticker_trading_data)
    
for name, df in trading_data_by_interval.items():
    print(f"=== {name} ===")
    display(df)
```

```python
output_dir = Path("../data/yfinance/")

for name, df in trading_data_by_interval.items():
    index_name = df.index.name  
    df = df.reset_index()    
    
    out_path = output_dir / f"{name}_interval_trading_data.csv"

    if out_path.exists():
        existing = pd.read_csv(out_path, parse_dates=[index_name])
        combined = pd.concat([existing, df], ignore_index=True)

        combined = combined.drop_duplicates(subset=[index_name, "Ticker"], keep="last")

    else:
        combined = df
    
    combined = combined.sort_values(["Ticker", index_name])

    combined.to_csv(out_path, index=False)
    print(f"Saved {len(combined)} total rows to {out_path} ({len(df)} rows fetched this run)")
```
